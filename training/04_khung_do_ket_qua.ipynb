# Khung đo · Bảng kết quả C0–C4

Gom mọi bản ghi `data/jobs/*/eval.json` (mỗi lần lồng tiếng ghi một bản) thành
bảng so sánh giữa các cấu hình. Đây là **Bảng 5.2** của đề cương, sinh ra từ số
đo thật chứ không gõ tay.

```python
# Chạy được dù mở notebook từ thư mục nào: đi ngược lên tới thư mục có backend/
import sys
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / "backend").is_dir() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
print("Gốc dự án:", ROOT)
```

```python
from backend.config import JOB_DIR
from backend.core import metrics

print("Các cấu hình đo:")
for cid, cfg in metrics.CONFIGS.items():
    print(f"  {cid}: {cfg.get('label', '')}")

records = metrics.collect(JOB_DIR)
print(f"\n{len(records)} lần chạy có bản ghi trong {JOB_DIR}")
if not records:
    print("Chưa có gì — chạy lồng tiếng ít nhất một lần rồi quay lại.")
```

## Bảng gom theo cấu hình

```python
rows = metrics.aggregate(records)
print(f"{len(records)} lần chạy · {len(rows)} cấu hình\n")
print(metrics.to_markdown(rows))
print()
for row in rows:
    label = metrics.CONFIGS.get(row["config"], {}).get("label", "")
    if label:
        print(f"  {row['config']}: {label}")
```

## Từng lần chạy

```python
import json

for r in records:
    print(json.dumps(r, ensure_ascii=False))
```

## Xuất CSV để dán vào báo cáo

```python
CSV_OUT = ROOT / "data" / "ket-qua-do.csv"
CSV_OUT.parent.mkdir(parents=True, exist_ok=True)
CSV_OUT.write_text(metrics.to_csv(rows), encoding="utf-8")
print("Đã ghi ->", CSV_OUT)
```
